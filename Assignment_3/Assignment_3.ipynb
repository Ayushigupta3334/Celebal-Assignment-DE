## Introduction

The Superstore dataset contains information about customer orders, products, sales, profits, and shipping details. In this project, SQL is used to analyze sales data and extract meaningful business insights. The dataset is first organized into separate tables for customers, orders, and products to improve data management and reduce redundancy.

## Objective

The objective of this project is to apply SQL concepts such as Subqueries, Common Table Expressions (CTEs), Window Functions, and Joins to analyze customer sales performance. The analysis aims to identify high-value customers, rank customers based on sales, evaluate purchasing behavior, and generate business insights from the Superstore dataset.


# Import required libraries

```python
import pandas as pd
import sqlite3
```

# Step 1: Setup Data
# Load the Superstore dataset

```python
df = pd.read_csv('superstore_raw.csv', encoding='latin1')
df.head()
```

# Create SQLite database connection

```python
conn = sqlite3.connect('superstore.db')
```

# Store the dataset into a SQL table named superstore_raw

```python
df.to_sql(
    'superstore_raw',
    conn,
    if_exists='replace',
    index=False
)
```

```python
pd.read_sql(
    "SELECT * FROM superstore_raw LIMIT 10",
    conn
)
```

```python
print(df.columns)
```

# Create customers table using distinct customer records

```python
conn.execute("""
CREATE TABLE customers AS
SELECT DISTINCT
    "Customer ID",
    "Customer Name",
    Segment
FROM superstore_raw
""")
```

```python
pd.read_sql("SELECT * FROM customers LIMIT 5", conn)
```

# Create products table using distinct product information

```python
conn.execute("""
CREATE TABLE products AS
SELECT DISTINCT
    "Product ID",
    "Product Name",
    Category,
    "Sub-Category"
FROM superstore_raw
""")
```

```python
pd.read_sql("SELECT * FROM products LIMIT 5", conn)
```

# Create orders table using distinct order details

```python
conn.execute("""
CREATE TABLE orders AS
SELECT DISTINCT
    "Order ID",
    "Order Date",
    "Ship Date",
    "Customer ID",
    Sales,
    Quantity,
    Profit
FROM superstore_raw
""")
```

```python
pd.read_sql("SELECT * FROM orders LIMIT 5", conn)
```

#Brief Insight :
The Superstore dataset was loaded into the superstore_raw table. Using SELECT DISTINCT, separate customer, product, and order tables were created to reduce duplication and organize the data into meaningful entities.



#Step 2: Performing Required Queries

# Query 1: Find orders with sales greater than average sales

```python
pd.read_sql("""
SELECT *
FROM orders
WHERE Sales > (
    SELECT AVG(Sales)
    FROM orders
)
""", conn)
```

#Insight:
These orders have sales values higher than the overall average sales, indicating above-average transactions.

# Query 2 : Find highest sales order for each customer

```python
pd.read_sql("""
SELECT *
FROM orders o
WHERE Sales = (
    SELECT MAX(Sales)
    FROM orders
    WHERE "Customer ID" = o."Customer ID"
)
""", conn)
```

#Insight:
This query identifies the largest purchase made by each customer.

# Query 3 : Calculate total sales for each customer using CTE

```python
pd.read_sql("""
WITH customer_sales AS (
    SELECT
        "Customer ID",
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY "Customer ID"
)
SELECT *
FROM customer_sales
ORDER BY total_sales DESC
""", conn)
```

#Insight:
The CTE calculates cumulative sales generated by each customer.

# Query 4 : Find customers whose total sales are above average


```python
pd.read_sql("""
WITH customer_sales AS (
    SELECT
        "Customer ID",
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY "Customer ID"
)
SELECT *
FROM customer_sales
WHERE total_sales >
(
    SELECT AVG(total_sales)
    FROM customer_sales
)
ORDER BY total_sales DESC
""", conn)
```

#Insight:
These customers contribute more sales than the average customer and are high-value customers.

# Query 5 : Rank customers based on total sales

```python
pd.read_sql("""
WITH customer_sales AS (
    SELECT
        "Customer ID",
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY "Customer ID"
)
SELECT
    "Customer ID",
    total_sales,
    RANK() OVER (ORDER BY total_sales DESC) AS sales_rank
FROM customer_sales
""", conn)
```

#Insight:
Customers are ranked according to their total contribution to sales revenue.

# Query 6 : Assign row numbers to orders within each customer

```python
pd.read_sql("""
SELECT
    "Customer ID",
    "Order ID",
    Sales,
    ROW_NUMBER() OVER (
        PARTITION BY "Customer ID"
        ORDER BY Sales DESC
    ) AS row_num
FROM orders
""", conn)
```

#Insight:
Each customer's orders are numbered separately based on sales value.

# Query 7 : Display top 3 customers based on total sales

```python
pd.read_sql("""
WITH customer_sales AS (
    SELECT
        "Customer ID",
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY "Customer ID"
),
ranked_customers AS (
    SELECT
        "Customer ID",
        total_sales,
        RANK() OVER (
            ORDER BY total_sales DESC
        ) AS sales_rank
    FROM customer_sales
)
SELECT *
FROM ranked_customers
WHERE sales_rank <= 3
""", conn)
```

#Insight:
These are the top-performing customers based on total sales generated.

#Brief Insight :
In this step, SQL techniques such as Subqueries, Common Table Expressions (CTEs), and Window Functions were used to analyze customer sales data. The queries helped identify high-value transactions, calculate total customer sales, determine customers with above-average performance, and rank customers based on revenue contribution. These analyses provide valuable insights into customer purchasing patterns and overall sales performance.

#Step 3: Final Combined Query

# Combine JOIN, CTE and Window Function
# Query : Display customer name, total sales and rank

```python
pd.read_sql("""
WITH customer_sales AS (
    SELECT
        "Customer ID",
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY "Customer ID"
)

SELECT
    c."Customer Name",
    cs.total_sales,
    RANK() OVER (
        ORDER BY cs.total_sales DESC
    ) AS sales_rank
FROM customer_sales cs
JOIN customers c
ON cs."Customer ID" = c."Customer ID"
ORDER BY sales_rank
""", conn)
```

#Brief Insight :
The query combines a CTE, JOIN, and Window Function to calculate total sales for each customer and rank them based on their contribution to revenue. Customers with higher total sales receive a better rank.

# MINI PROJECT

# Query 1 :Find the top 5 customers based on total sales generated

```python
pd.read_sql("""
WITH customer_sales AS (
    SELECT
        "Customer ID",
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY "Customer ID"
)

SELECT
    c."Customer Name",
    cs.total_sales
FROM customer_sales cs
JOIN customers c
ON cs."Customer ID" = c."Customer ID"
ORDER BY cs.total_sales DESC
LIMIT 5
""", conn)
```

#Insight

These customers generated the highest revenue and can be considered the most valuable customers.

# Query 2 : Find the bottom 5 customers based on total sales

```python
pd.read_sql("""
WITH customer_sales AS (
    SELECT
        "Customer ID",
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY "Customer ID"
)

SELECT
    c."Customer Name",
    cs.total_sales
FROM customer_sales cs
JOIN customers c
ON cs."Customer ID" = c."Customer ID"
ORDER BY cs.total_sales ASC
LIMIT 5
""", conn)
```

#Insight

These customers contributed the least sales and may require engagement strategies to increase purchases.

# Query 3 : Identify customers who placed only one order

```python
pd.read_sql("""
SELECT
    c."Customer Name",
    COUNT(o."Order ID") AS order_count
FROM customers c
JOIN orders o
ON c."Customer ID" = o."Customer ID"
GROUP BY c."Customer ID", c."Customer Name"
HAVING COUNT(o."Order ID") = 1
""", conn)
```

#Insight

These customers placed only one order during the period and may represent one-time buyers.

# Query 4 : Find customers whose total sales are greater than the average customer sales

```python
pd.read_sql("""
WITH customer_sales AS (
    SELECT
        "Customer ID",
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY "Customer ID"
)

SELECT
    c."Customer Name",
    cs.total_sales
FROM customer_sales cs
JOIN customers c
ON cs."Customer ID" = c."Customer ID"
WHERE cs.total_sales >
(
    SELECT AVG(total_sales)
    FROM customer_sales
)
ORDER BY cs.total_sales DESC
""", conn)
```

#Insight

These customers spend more than the average customer and are important contributors to revenue.

# Query 5 : Find the highest order value placed by each customer

```python
pd.read_sql("""
SELECT
    c."Customer Name",
    MAX(o.Sales) AS highest_order_value
FROM customers c
JOIN orders o
ON c."Customer ID" = o."Customer ID"
GROUP BY c."Customer ID", c."Customer Name"
ORDER BY highest_order_value DESC
""", conn)
```

#Insight

This shows the maximum value of a single order placed by each customer.
