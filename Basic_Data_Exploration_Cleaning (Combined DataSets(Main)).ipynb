As the dataset did not include a quantity column, a quantity column was created and used to calculate total_amount.

```python
df['quantity'] = 2
df['total_amount'] = df['final_price'] * df['quantity']
```

```python

df[['final_price', 'quantity', 'total_amount']].head()
```

```python
df[['final_price', 'quantity']].dtypes
```

The final_price column was originally stored as a string/object data type because it contained currency symbols (₹) and commas. To perform mathematical operations and create the total_amount column, the unwanted symbols were removed and the data type was converted to float. This allowed accurate numerical calculations using the price values.

```python
df['final_price'] = df['final_price'].str.replace('₹', '', regex=False)

df['final_price'] = df['final_price'].str.replace('"', '', regex=False)

df['final_price'] = df['final_price'].str.replace(',', '', regex=False)

df['final_price'] = df['final_price'].astype(float)
```

```python
df['final_price'].head()
```

# Derived Column

```python
df['total_amount'] = df['final_price'] * df['quantity']
```

```python
df[['final_price', 'quantity', 'total_amount']].head()
```
