# Q7: How does Spark use the Lineage Graph (DAG) to provide fault tolerance if a worker node fails?

Spark stores transformations in a DAG (Lineage Graph). If a node fails, Spark recomputes only the lost partition instead of the whole dataset.

# Q9: Explain the concept of Predicate Pushdown in Parquet and how it affects the amount of data loaded into memory.

Predicate Pushdown allows Spark to apply filter conditions while reading Parquet files, so only the required rows are loaded into memory. This reduces I/O and improves performance.

# Q11: What is the difference between Transformations and Actions? Provide two examples of each.

Transformations: Return a new DataFrame and are lazily evaluated.

Examples:

1. select()
2. filter()

Actions: Execute the transformations and produce a result.

Examples:

1. show()
2. collect()

# Q12: Write the Spark command to load a Parquet file from "path/to/input", filter out any rows where user_id is null, and save the result as a CSV at "path/to/output".

```python
df.write.mode("overwrite").parquet("input_parquet")
```

```python
parquet_df = spark.read.parquet("input_parquet")
```

```python
result = parquet_df.filter(col("Customer ID").isNotNull())
```

```python
result.write.mode("overwrite").option("header", True).csv("output_csv")
```

# Q15: When exploring a dataset, why is it safer to use .show(5) instead of .collect() on a multi-terabyte dataset?

.show(5) displays only the first five rows, making it fast and memory-efficient.

.collect() retrieves all rows to the driver, which can cause memory issues with very large datasets. Therefore, .show(5) is preferred for exploring large datasets.
