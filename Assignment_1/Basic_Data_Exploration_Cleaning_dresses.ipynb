```python
df.ffill(inplace=True)
```

```python
df.dropna(inplace=True)
```

```python
df[['final_price', 'discount']]
```

```python
high_discount = df[df['discount'] > 70.0]
high_discount
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['quantity'] = 3
df['total_amount'] = df['final_price'] * df['quantity']
```

```python
df['final_price'] = df['final_price'].str.replace('₹', '', regex=False)

df['final_price'] = df['final_price'].str.replace('"', '', regex=False)

df['final_price'] = df['final_price'].str.replace(',', '', regex=False)

df['final_price'] = df['final_price'].astype(float)
```

```python
df['final_price'].head()
```

```python
df['total_amount'] = df['final_price'] * df['quantity']
```

```python
df[['final_price', 'quantity', 'total_amount']].head()
```

```python
df.to_csv("cleaned_dataset(dresses).csv", index=False)
```
